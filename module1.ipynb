```python
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """
        output = []
        for index, num in enumerate(nums):
            for indexCheck, numCheck in enumerate(nums):
                if indexCheck != index and (num+numCheck) == target:
                    if index not in output:
                        output.append(index)
                    if indexCheck not in output:
                        output.append(indexCheck)
                  
        return output
                        

result = Solution()

print(result.twoSum([2, 7, 11, 15], 9))
```

```python
class Solution(object):
    def addTwoNumbers(self, l1, l2):
        """
        :type l1: ListNode
        :type l2: ListNode
        :rtype: ListNode
        """
        output = []
        memory = 0
        for index in range(len(l1)):
            flag = False
            if index >= len(l2):
                flag = True
            
            if flag:
                if (l1[index] + memory) % 10 == 0 and (l1[index] + memory) > 10:
                    memory = int(str(l1[index] + memory)[0])
                    output.append(int(str(l1[index]  + memory)[-1]))
                elif (l1[index] + memory) % 10 == 0:
                    memory = int(str(l1[index] + memory)[0])
                    output.append(0)
                elif (l1[index] + memory) < 10:
                    output.append(l1[index]  + memory)
            else:
                if l1[index] == 0 and l2[index] == 0:
                    output.append(0)
                elif (l1[index] + l2[index] + memory) % 10 == 0 and (l1[index] + l2[index] + memory) > 10:
                    memory = int(str(l1[index] + l2[index] + memory)[0])
                    output.append(int(str(l1[index] + l2[index] + memory)[-1]))
                elif (l1[index] + l2[index] + memory) % 10 == 0:
                    memory = int(str(l1[index] + l2[index] + memory)[0])
                    output.append(0)
                elif (l1[index] + l2[index] + memory) < 10:
                    output.append(l1[index] + l2[index] + memory)
                    memory = 0
        return output
        
output = Solution()
print(output.addTwoNumbers([9,9,9,9,9,9,9], [9,9,9,9]))
```

```python
class Solution(object):
    def lengthOfLongestSubstring(self, s):
        """
        :type s: str
        :rtype: int
        """
        strLongest = ""
        strTemp = ""
        for index in range(len(s)):
            for char in s:
                if char not in strTemp:
                    strTemp += char
                if len(strTemp) > len(strLongest):
                    strLongest = strTemp
                
        return f'The answer is "{strLongest}", with the length of {len(strLongest)}.'
            
```
